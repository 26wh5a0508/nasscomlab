```python
#histogram
import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")

plt.hist(
    tips["total_bill"],
    bins=30,
    edgecolor="black"
)

plt.show()
```

```python
#pie chart
counts = tips["smoker"].value_counts()
plt.pie(
counts,
labels=counts.index,
autopct="%1.1f%%",
startangle=80
)
```

```python
#scatter plot
import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")

plt.scatter(
    tips["total_bill"],
    tips["tip"]
)

plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.title("Total Bill vs Tip")

plt.show()
```

```python
#bar plot
import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")

avg_bill = tips.groupby("day")["total_bill"].mean()

plt.bar(avg_bill.index, avg_bill.values, edgecolor="black")

plt.xlabel("Day")
plt.ylabel("Average Total Bill")
plt.title("Average Total Bill by Day")
plt.show()
```

```python
#horizontal bars
import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")

avg_bill = tips.groupby("day")["total_bill"].mean()

plt.barh(
    avg_bill.index,
    avg_bill.values,
    edgecolor="black"
)

plt.xlabel("Average Total Bill")
plt.ylabel("Day")
plt.title("Average Total Bill by Day")

plt.show()
```

```python
#vertical bar
plt.bar(
    avg_bill.index,
    avg_bill.values,
    edgecolor="black"
)

plt.xlabel("Day")
plt.ylabel("Average Total Bill")
plt.title("Average Total Bill by Day")

plt.show()
```

```python
#box plot
import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")

plt.boxplot(tips["total_bill"])

plt.ylabel("Total Bill")
plt.title("Box Plot of Total Bill")

plt.show()
```

```python
#count plot
sns.countplot(
    data=tips,
    x="day",
    color="yellow"
)
```

```python
#swarm plot
sns.swarmplot(
    data=tips,
    x="day",
    y="total_bill",
    color="orange"
)

plt.show()
```

```python
sns.swarmplot(
    data=tips,
    x="day",
    y="tip",
    color="yellow"
)

plt.xlabel("Day")
plt.ylabel("Tip")

plt.show()
```

```python
#final code
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="smoker"
)
```
