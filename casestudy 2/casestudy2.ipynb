```python
import pandas as pd
import numpy as np

# Create the dataset
data = {
    "Student": ["S1", "S2", "S3", "S4", "S5", "S6", "S7", "S8", "S9", "S10"],
    "Study_Hours": [2, 3, 4, 5, 6, 7, 8, 4, 6, 9],
    "Attendance": [65, 70, 75, 80, 85, 90, 92, 78, 88, 95],
    "Assignment_Score": [55, 60, 65, 72, 78, 85, 88, 70, 80, 92],
    "Previous_Score": [58, 62, 68, 70, 76, 82, 86, 72, 79, 90],
    "Final_Score": [60, 64, 70, 75, 82, 88, 91, 74, 84, 94]
}

# Create DataFrame
df = pd.DataFrame(data)

# Select input variables
X = df[[
    "Study_Hours",
    "Attendance",
    "Assignment_Score"
]].values

# Select output variable
Y = df["Final_Score"].values

# Add column of 1s for intercept
X = np.column_stack((np.ones(len(X)), X))

# Calculate coefficients
beta = np.linalg.inv(X.T @ X) @ X.T @ Y

# Display results
print("Coefficients:")
print(beta)

print("\nIntercept (b0):", beta[0])
print("Study Hours coefficient (b1):", beta[1])
print("Attendance coefficient (b2):", beta[2])
print("Assignment Score coefficient (b3):", beta[3])
```

```python
average_final_score = df["Final_Score"].mean()
variation = df["Final_Score"].std()

highest_student = df.loc[df["Final_Score"].idxmax(), "Student"]
highest_score = df["Final_Score"].max()

print("Average Final Score:", average_final_score)
print("Standard Deviation:", variation)
print("Highest Scoring Student:", highest_student)
print("Highest Final Score:", highest_score)

print("\nCorrelation with Final Score:")
print(df.corr(numeric_only=True)["Final_Score"])
```

```python
The average final score of the students is 78.2.
 The final scores show some variation around this average, as indicated by their standard deviation.
 S10 obtained the highest final score of 94, while S1 obtained the lowest score of 60.
 The data shows a generally positive relationship between study hours, attendance, and final score.
Correlation analysis can be used to identify which input variable has the strongest relationship with Final Score.
Matrix representation makes it possible to analyze several input variables together, while differentiation describes the change in the predicted score with respect to study hours.
The integration result represents the total learning benefit accumulated over the specified range of study hours.
```
